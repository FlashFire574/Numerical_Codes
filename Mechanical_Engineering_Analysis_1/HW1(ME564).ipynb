```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

Question 2b)

```python
#Given parameters
x0=10
t=np.linspace(0,5,300)
lambdas=[-5, -1, 0, 0.01, 0.1]

# Plot each lambda
plt.figure(figsize=(8, 5))
for lam in lambdas:
    x = x0 * np.exp(lam * t)
    plt.plot(t, x, label=f'lambda={lam}')

# Labels and legend
plt.title("Radioactive Decay/Growth for Different λ Values")
plt.xlabel("Time t")
plt.ylabel("Mass x(t)")
plt.legend()
plt.grid(True)
plt.show()
```

Question 3a) and 3b)

```python
# Define range
x = np.linspace(-5, 5, 500)

# Define functions
def f(x):
    return np.sin(x)/x

def g(x):
    return (np.cos(x) - 1)/x

# Taylor expansions
f_taylor = 1 - x**2/6 + x**4/120
g_taylor = -x/2 + x**3/24 - x**5/720

# Plot f(x) = sin(x)/x
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(x, f(x), label='f(x) = sin(x)/x')
plt.plot(x, f_taylor, '--', label='f(x)_taylor')
plt.title('f(x) = sin(x)/x')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)

# Plot g(x) = (cos(x) - 1)/x
plt.subplot(1, 2, 2)
plt.plot(x, g(x), label='g(x) = (cos(x)-1)/x')
plt.plot(x, g_taylor, '--', label='g(x)_taylor')
plt.title('g(x) = (cos(x) - 1)/x')
plt.xlabel('x')
plt.ylabel('g(x)')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

Question 4a), 4b) and 4c)

```python
# Time range
t = np.arange(0, 10.01, 0.01)

# 4(a)
f_a_real = np.cos(t)
f_a_imag = np.sin(t)

# 4(b)
f_b_real = np.exp(t) * np.cos(t)
f_b_imag = -np.exp(t) * np.sin(t)

# 4(c)
# f_c_real = np.exp(-t) * np.cos(np.pi * t)
# f_c_imag = -np.exp(-t) * np.sin(np.pi * t)
f_c=np.exp(-t) * np.exp(-1j * np.pi * t)
f_c_real = np.real(f_c)
f_c_imag = np.imag(f_c)

# Plot all three
fig, axs = plt.subplots(3, 1, figsize=(8, 10))

# (a)
axs[0].plot(t, f_a_real, label='Real part')
axs[0].plot(t, f_a_imag, '--', label='Imag part')
axs[0].set_xlabel('Time t')
axs[0].set_ylabel('f(t)')
axs[0].set_title('f(t) = e^{it}')
axs[0].legend()
axs[0].grid(True)

# (b)
axs[1].plot(t, f_b_real, label='Real part')
axs[1].plot(t, f_b_imag, '--', label='Imag part')
axs[1].set_xlabel('Time t')
axs[1].set_ylabel('f(t)')
axs[1].set_title('f(t) = e^{(1 - i)t}')
axs[1].legend()
axs[1].grid(True)

# (c)
axs[2].plot(t, f_c_real, label='Real part')
axs[2].plot(t, f_c_imag, '--', label='Imag part')
axs[2].set_xlabel('Time t')
axs[2].set_ylabel('f(t)')
axs[2].set_title('f(t) = e^{(-1 - πi)t}')
axs[2].legend()
axs[2].grid(True)

plt.tight_layout()
plt.show()
```

3D Plots

```python
# Time range
t = np.arange(0, 10.01, 0.01)

# 4(a)
f_a_real = np.cos(t)
f_a_imag = np.sin(t)

# 4(b) f(t) = e^{(1 - i)t}
f_b_real = np.exp(t) * np.cos(t)
f_b_imag = -np.exp(t) * np.sin(t)

# 4(c) f(t) = e^{(-1 - πi)t}
f_c_real = np.exp(-t) * np.cos(np.pi * t)
f_c_imag = -np.exp(-t) * np.sin(np.pi * t)

# Setup figure
fig = plt.figure(figsize=(20, 20))
plt.subplots_adjust(hspace=0.5)

# (a)
ax1 = fig.add_subplot(3, 1, 1, projection='3d')
ax1.plot(f_a_real, f_a_imag, t, color='b')
ax1.set_title('a) f(t) = e^(it)')
ax1.set_xlabel('Re')
ax1.set_ylabel('Im')
ax1.set_zlabel('t')
ax1.grid(True)

# (b)
ax2 = fig.add_subplot(3, 1, 2, projection='3d')
ax2.plot(f_b_real, f_b_imag, t, color='r')
ax2.set_title('b) f(t) = e^(1 - i)t')
ax2.set_xlabel('Re')
ax2.set_ylabel('Im')
ax2.set_zlabel('t')
ax2.grid(True)

# (c)
ax3 = fig.add_subplot(3, 1, 3, projection='3d')
ax3.plot(f_c_real, f_c_imag, t, color='g')
ax3.set_title('c) f(t) = e^(-1 - πi)t')
ax3.set_xlabel('Re')
ax3.set_ylabel('Im')
ax3.set_zlabel('t')
ax3.grid(True)

plt.tight_layout()
plt.show()
```
